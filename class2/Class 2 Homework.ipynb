## 📝 Tesseract OCR Guide

Tesseract is an open-source Optical Character Recognition (OCR) engine developed by Google. It is used to convert scanned images, PDFs, and images with text into machine-readable text. Tesseract supports more than 100 languages, including complex scripts like Arabic, Chinese, and many others.

---

## 🛠️ Basic Usage

Once Tesseract is installed, you can use it either from the command line or by using the Python wrapper `pytesseract`.

### **Command Line Usage**

To perform OCR on an image:
```bash
tesseract image.png output.txt


## Use Tesseract with python

````python
from PIL import Image
import pytesseract #pip install pytesseract first

# Load an image using Pillow (PIL)
image = Image.open('image.png')

# Perform OCR on the image
text = pytesseract.image_to_string(image)

print(text)
````

Tesseract supports over 100 languages, and you can even train it for custom languages or fonts. To use a different language, you can download the corresponding trained data files and specify the language in the -l flag.

For example, to use Spanish (spa):

````python
tesseract image.png output -l spa
````

🧑‍💻 Tesseract Best Practices\
Preprocess Images: Always preprocess images by converting to grayscale, adjusting brightness/contrast, and removing noise to improve Tesseract’s accuracy.

Use Correct --psm: The Page Segmentation Mode (--psm) plays a crucial role in how Tesseract segments the image and interprets the text. Experiment with different modes for complex documents.

Choose the Right Language: Always specify the correct language (-l lang_code) for better accuracy. Tesseract performs poorly when the language is incorrect.

Custom Training: For specialized fonts or languages, you can train Tesseract to recognize custom fonts or languages. This is especially useful for handwriting or unusual fonts.

## Task 2 — PDF → Text + Native Image Extraction

### Why not Tesseract?

Tesseract is an **image OCR** engine — it converts screenshots or scanned pages into text.  
Our PDFs are **text-based** (not scanned), so parsing the PDF structure directly is both faster and more accurate.

| | Tesseract OCR | MinerU + PyMuPDF (this approach) |
|---|---|---|
| Text quality | OCR errors, no layout awareness | Clean, layout-aware parsing |
| Tables | OCR guesswork | Structured HTML |
| Images | Screenshot crop | Native PDF binary extraction |
| Formulas | OCR fails | MinerU LaTeX-aware |

### Pipeline (see `hw_output/pdf_ocr/pipeline_overview.md` for full Mermaid diagram)

```
PDF
 ├─ MinerU  →  {doc_id}.md  +  content_list.json  +  images/
 └─ PyMuPDF →  native images (p{page}_i{idx}_xref{id}.jpeg)
                     ↓
              build_sidecar.py
                     ↓
              papers_corpus/{doc_id}/
                ├─ manifest.json            (figure/table inventory)
                ├─ paper_native_images.md   (final markdown with native images)
                └─ tables_structured.md     (extracted tables)
```

````python
import pathlib, json

sample_dir = pathlib.Path("hw_output/pdf_ocr/sample/chang-2023")

# Show text extraction result (first 60 lines of MinerU markdown)
md_lines = (sample_dir / "chang-2023.md").read_text().splitlines()
print("=== MinerU Text Output (first 30 lines of chang-2023.md) ===\n")
print("\n".join(md_lines[:30]))

# Show extracted native images
native_imgs = sorted((sample_dir / "native_images").glob("*.*"))
print(f"\n\n=== PyMuPDF Native Images Extracted: {len(native_imgs)} files ===")
for img in native_imgs:
    size_kb = img.stat().st_size / 1024
    print(f"  {img.name:<40} {size_kb:5.1f} KB")

# Show content_list structure
cl = json.loads((sample_dir / "chang-2023_content_list.json").read_text())
types = {}
for block in cl:
    t = block.get("type", "unknown")
    types[t] = types.get(t, 0) + 1
print(f"\n=== content_list.json block types ===")
for k, v in sorted(types.items(), key=lambda x: -x[1]):
    print(f"  {k:<20} {v:>4} blocks")
````
